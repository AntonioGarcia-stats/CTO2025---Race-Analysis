```python
%pip install pdfplumber
import pdfplumber 
import pandas as pd

pdf_path = r"C:\Users\garci\OneDrive\Desktop\CTO2025RESULTS.PDF"

rows = []

with pdfplumber.open(pdf_path) as pdf:
    for page in pdf.pages:
        table = page.extract_table()

        if table:
            rows.extend(table)
df = pd.DataFrame(rows)
df.head()


df.shape
df.head(15)
df.tail(15)
df.info()

df.columns

columns = [ 
    "OVR",
    "NAME",
    "HOMETOWN",
    "DIVISION",
    "TIME",
    "PACE",
    "GENDER_RANK",
    "DIV_RANK",
    "BIB"
]

df = df.iloc[:, :9]
df.columns = columns

df = df[df["OVR"].notna()]
df = df[df["OVR"] != "OVR"]

df.head(20)

numeric_cols = ["OVR", "GENDER_RANK", "DIV_RANK", "BIB"]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df["GENDER"] = df["DIVISION"].str.extract(r"^(Male|Female)")

df["AGE_GROUP"] = (
    df["DIVISION"]
    .str.replace("Male ", "", regex=False)
    .str.replace("Female ", "", regex=False)
)


def time_to_minutes(x):
    try:
        h, m, s = str(x).split(":")
        return float(h) * 60 + float(m) + float(s) / 60
    except:
        return None

df["TIME_MINUTES"] = df["TIME"].apply(time_to_minutes)


df.head()
df.info()
df["TIME_MINUTES"].describe()


df.to_csv("CTO2025_clean.csv", index=False)
df.to_csv("C:/Users/garci/OneDrive/Desktop/CTO2025_clean.csv", index=False)
```
