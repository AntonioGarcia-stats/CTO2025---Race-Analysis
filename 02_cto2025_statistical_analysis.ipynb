```python
# CONQUER THE OVERLOOK 2025 - STATISTICAL ANALYSIS


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats
import statsmodels.api as sm
import statsmodels.formula.api as smf



# 1. LOAD CLEANED DATA
# ---------------------

df = pd.read_csv("/CTO2025_clean.csv")

print("Dataset shape:", df.shape)
display(df.head())
df.info()


# 2. PREPARE VARIABLES
# --------------------

age_order = [
    "19 & Under",
    "20-29",
    "30-39",
    "40-49",
    "50-59",
    "60-69",
    "70-79",
    "80-89"
]

df["AGE_GROUP"] = pd.Categorical(
    df["AGE_GROUP"],
    categories=age_order,
    ordered=True
)

analysis_df = df.dropna(
    subset=["TIME_MINUTES", "AGE_GROUP", "GENDER"]
).copy()


# 3. FREQUENCY OF RUNNERS BY HOMETOWN
# ------------------------------------

hometown_counts = df["HOMETOWN"].value_counts()

plt.figure(figsize=(12, 6))

bars = plt.bar(
    hometown_counts.index,
    hometown_counts.values
)

plt.title("Frequency of Runners by Hometown")
plt.xlabel("Hometown")
plt.ylabel("Number of Runners")
plt.xticks(rotation=45, ha="right")

for bar in bars:
    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        str(int(height)),
        ha="center",
        va="bottom",
        fontsize=8
    )

plt.tight_layout()
plt.show()



# 4. AGE GROUP FREQUENCY STRATIFIED BY GENDER
# -------------------------------------------

age_gender_counts = pd.crosstab(
    analysis_df["AGE_GROUP"],
    analysis_df["GENDER"]
)

print("\nAge Group by Gender Frequency Table")
print(age_gender_counts)

ax = age_gender_counts.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Frequency of Age Group Stratified by Gender")
plt.xlabel("Age Group")
plt.ylabel("Number of Runners")
plt.xticks(rotation=45)
plt.legend(title="Gender")

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.show()


# 5. OVERALL DESCRIPTIVE STATISTICS
# ---------------------------------

overall_stats = pd.DataFrame({
    "N": [analysis_df["TIME_MINUTES"].count()],
    "Mean": [analysis_df["TIME_MINUTES"].mean()],
    "SD": [analysis_df["TIME_MINUTES"].std()],
    "Median": [analysis_df["TIME_MINUTES"].median()],
    "Min": [analysis_df["TIME_MINUTES"].min()],
    "Q1": [analysis_df["TIME_MINUTES"].quantile(0.25)],
    "Q3": [analysis_df["TIME_MINUTES"].quantile(0.75)],
    "Max": [analysis_df["TIME_MINUTES"].max()]
})

overall_stats["IQR"] = (
    overall_stats["Q3"] - overall_stats["Q1"]
)

print("\nOverall Finish Time Statistics")
display(overall_stats.round(2))


# 6. DESCRIPTIVE STATISTICS BY AGE GROUP
# --------------------------------------

age_stats = (
    analysis_df
    .groupby("AGE_GROUP", observed=True)["TIME_MINUTES"]
    .agg(
        N="count",
        Mean="mean",
        SD="std",
        Median="median",
        Min="min",
        Q1=lambda x: x.quantile(0.25),
        Q3=lambda x: x.quantile(0.75),
        Max="max"
    )
)

age_stats["IQR"] = age_stats["Q3"] - age_stats["Q1"]

print("\nFinish Time Statistics by Age Group")
display(age_stats.round(2))


# 7. DESCRIPTIVE STATISTICS BY GENDER
# -------------------------------------

gender_stats = (
    analysis_df
    .groupby("GENDER")["TIME_MINUTES"]
    .agg(
        N="count",
        Mean="mean",
        SD="std",
        Median="median",
        Min="min",
        Q1=lambda x: x.quantile(0.25),
        Q3=lambda x: x.quantile(0.75),
        Max="max"
    )
)

gender_stats["IQR"] = (
    gender_stats["Q3"] - gender_stats["Q1"]
)

print("\nFinish Time Statistics by Gender")
display(gender_stats.round(2))

# 8. DESCRIPTIVE STATISTICS BY AGE GROUP AND GENDER
# --------------------------------------------------

age_gender_stats = (
    analysis_df
    .groupby(
        ["AGE_GROUP", "GENDER"],
        observed=True
    )["TIME_MINUTES"]
    .agg(
        N="count",
        Mean="mean",
        SD="std",
        Median="median",
        Min="min",
        Q1=lambda x: x.quantile(0.25),
        Q3=lambda x: x.quantile(0.75),
        Max="max"
    )
)

age_gender_stats["IQR"] = (
    age_gender_stats["Q3"] -
    age_gender_stats["Q1"]
)

print("\nFinish Time Statistics by Age Group and Gender")
display(age_gender_stats.round(2))


# 9. BOXPLOT OF FINISH TIME BY AGE GROUP AND GENDER
# --------------------------------------------------

male_data = []
female_data = []

for age in age_order:

    male_times = analysis_df.loc[
        (analysis_df["AGE_GROUP"] == age) &
        (analysis_df["GENDER"] == "Male"),
        "TIME_MINUTES"
    ]

    female_times = analysis_df.loc[
        (analysis_df["AGE_GROUP"] == age) &
        (analysis_df["GENDER"] == "Female"),
        "TIME_MINUTES"
    ]

    male_data.append(male_times)
    female_data.append(female_times)


x = np.arange(len(age_order))

male_positions = x - 0.2
female_positions = x + 0.2

fig, ax = plt.subplots(figsize=(12, 7))

bp1 = ax.boxplot(
    male_data,
    positions=male_positions,
    widths=0.35,
    patch_artist=True,
    manage_ticks=False
)

bp2 = ax.boxplot(
    female_data,
    positions=female_positions,
    widths=0.35,
    patch_artist=True,
    manage_ticks=False
)

for box in bp1["boxes"]:
    box.set_facecolor("lightblue")

for box in bp2["boxes"]:
    box.set_facecolor("lightpink")

y_min = analysis_df["TIME_MINUTES"].min()

ax.set_ylim(bottom=y_min - 15)

for i in range(len(age_order)):

    ax.text(
        male_positions[i],
        y_min - 7,
        f"n={len(male_data[i])}",
        ha="center",
        va="top",
        fontsize=9
    )

    ax.text(
        female_positions[i],
        y_min - 7,
        f"n={len(female_data[i])}",
        ha="center",
        va="top",
        fontsize=9
    )

ax.set_xticks(x)
ax.set_xticklabels(age_order, rotation=45)

ax.set_xlabel("Age Group")
ax.set_ylabel("Finish Time (Minutes)")
ax.set_title(
    "Finish Times Stratified by Age Group and Gender"
)

ax.plot(
    [],
    [],
    color="lightblue",
    linewidth=8,
    label="Male"
)

ax.plot(
    [],
    [],
    color="lightpink",
    linewidth=8,
    label="Female"
)

ax.legend(title="Gender")

plt.tight_layout()
plt.show()


# 10. COHEN'S D: MALE VS FEMALE
# ------------------------------

male_times = analysis_df.loc[
    analysis_df["GENDER"] == "Male",
    "TIME_MINUTES"
]

female_times = analysis_df.loc[
    analysis_df["GENDER"] == "Female",
    "TIME_MINUTES"
]

n_male = len(male_times)
n_female = len(female_times)

mean_male = male_times.mean()
mean_female = female_times.mean()

sd_male = male_times.std(ddof=1)
sd_female = female_times.std(ddof=1)

pooled_sd = np.sqrt(
    (
        (n_male - 1) * sd_male**2 +
        (n_female - 1) * sd_female**2
    )
    /
    (n_male + n_female - 2)
)

cohens_d = (
    mean_male - mean_female
) / pooled_sd

print("\nCohen's d - Male vs Female")

print(
    "Male Mean:",
    round(mean_male, 2)
)

print(
    "Female Mean:",
    round(mean_female, 2)
)

print(
    "Cohen's d:",
    round(cohens_d, 3)
)


# 11. ETA-SQUARED: AGE GROUP
# ---------------------------

overall_mean = analysis_df[
    "TIME_MINUTES"
].mean()

ss_total = (
    (
        analysis_df["TIME_MINUTES"] -
        overall_mean
    ) ** 2
).sum()

ss_between = 0

for age, group in analysis_df.groupby(
    "AGE_GROUP",
    observed=True
):

    group_times = group[
        "TIME_MINUTES"
    ]

    ss_between += (
        len(group_times) *
        (group_times.mean() - overall_mean) ** 2
    )

eta_squared = (
    ss_between / ss_total
)

print(
    "\nEta-squared for Age Group:",
    round(eta_squared, 3)
)


# 12. OLS REGRESSION MODEL
# -------------------------

reg_df = analysis_df.copy()

model = smf.ols(
    """
    TIME_MINUTES ~
    C(AGE_GROUP, Treatment(reference='20-29')) +
    C(GENDER, Treatment(reference='Female'))
    """,
    data=reg_df
).fit()

print("\nOLS Regression Results")
print(model.summary())


# 13. ANOVA TABLE
# ----------------

anova_results = sm.stats.anova_lm(
    model,
    typ=2
)

print("\nANOVA Results")
display(anova_results.round(4))


# 14. PARTIAL ETA-SQUARED
# ------------------------

residual_ss = anova_results.loc[
    "Residual",
    "sum_sq"
]

anova_results["partial_eta_sq"] = np.nan

for effect in anova_results.index:

    if effect != "Residual":

        effect_ss = anova_results.loc[
            effect,
            "sum_sq"
        ]

        anova_results.loc[
            effect,
            "partial_eta_sq"
        ] = (
            effect_ss /
            (effect_ss + residual_ss)
        )

print("\nANOVA with Partial Eta-Squared")
display(anova_results.round(4))


# 15. MODEL DIAGNOSTICS
# ---------------------

# Residuals vs Fitted
plt.figure(figsize=(8, 6))

plt.scatter(
    model.fittedvalues,
    model.resid
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Fitted Finish Time")
plt.ylabel("Residuals")
plt.title("Residuals vs Fitted Values")

plt.tight_layout()
plt.show()


# Q-Q Plot
plt.figure(figsize=(8, 6))

stats.probplot(
    model.resid,
    dist="norm",
    plot=plt
)

plt.title("Q-Q Plot of Regression Residuals")

plt.tight_layout()
plt.show()


# 16. MODEL FIT SUMMARY
# ---------------------

print("\nModel Fit")

print(
    "R-squared:",
    round(model.rsquared, 3)
)

print(
    "Adjusted R-squared:",
    round(model.rsquared_adj, 3)
)

print(
    "Number of observations:",
    int(model.nobs)
)


# 17. FINAL SAMPLE COUNTS
# ------------------------

print("\nGender Counts")
print(
    analysis_df["GENDER"].value_counts()
)

print("\nAge Group Counts")
print(
    analysis_df["AGE_GROUP"]
    .value_counts()
    .sort_index()
)

print("\nAge Group × Gender Counts")
print(
    pd.crosstab(
        analysis_df["AGE_GROUP"],
        analysis_df["GENDER"]
    )
)
```
